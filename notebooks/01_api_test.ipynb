```python
# ==========================================
# Configuration
# ==========================================
```

```python
API_URL = "https://data.cityofnewyork.us/resource/erm2-nwe9.json"

AGENCY = "HPD"

COMPLAINT_TYPE = "HEAT/HOT WATER"
```

```python
# ==========================================
# Selected Columns
# ==========================================
```

```python
selected_columns = [
    "unique_key",
    "created_date",
    "closed_date",
    "agency",
    "agency_name",
    "complaint_type",
    "descriptor",
    "descriptor_2",
    "location_type",
    "incident_zip",
    "incident_address",
    "street_name",
    "status",
    "due_date",
    "resolution_description",
    "resolution_action_updated_date",
    "borough",
    "community_board",
    "council_district",
    "bbl",
    "open_data_channel_type",
    "latitude",
    "longitude"
]
# Fields required for operational reporting, geographic analysis,
# complaint tracking, and Power BI dashboards.
```

# Filtered HPD Heat/Hot Water API Pull

```python
selected_columns_string = ",".join(selected_columns)

params = {
    "$select": selected_columns_string,
    "$where": (
        f"agency = '{AGENCY}' "
        f"AND complaint_type = '{COMPLAINT_TYPE}'"
    ),
    "$order": "created_date DESC",
    "$limit": 5
}

response = requests.get(
    API_URL,
    params=params,
    timeout=30
)

response.raise_for_status()

records = response.json()

print("HTTP status:", response.status_code)
print("Rows returned:", len(records))
print("Request URL:")
print(response.url)
```

```python
records[0]
```

```python
heat_df = pd.DataFrame(records)

print("Shape:", heat_df.shape)
display(heat_df)
```

```python
print("Agencies:")
display(heat_df["agency"].value_counts(dropna=False))

print("Complaint types:")
display(heat_df["complaint_type"].value_counts(dropna=False))
```

```python
print("Requested columns:", len(selected_columns))
print("Returned columns:", len(heat_df.columns))

missing_from_response = [
    column
    for column in selected_columns
    if column not in heat_df.columns
]

print("Fields absent from this five-row response:")
print(missing_from_response)
```

```python
heat_df = heat_df.reindex(columns=selected_columns)

display(heat_df)
```

```python
print("Shape:", heat_df.shape)
```

```python
assert response.status_code == 200
assert len(records) == 5
assert heat_df["agency"].dropna().eq("HPD").all()
assert heat_df["complaint_type"].dropna().eq("HEAT/HOT WATER").all()
assert heat_df["unique_key"].notna().all()

print("Filtered API validation passed.")
```
